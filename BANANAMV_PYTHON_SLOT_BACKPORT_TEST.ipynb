```python
#@title ① 安裝 CLASSIC Python Slot 5070 Python 移植測試分支、ComfyUI 與外掛
import os, sys, subprocess, shutil
from pathlib import Path
PROJECT = Path('/content/banana-mv-classic-python-slot')
BRANCH = 'experiment/colab-python-slot-5070-backport'
REPOSITORY = 'https://github.com/haohao1130-bit/banana-mv-public.git'
if not PROJECT.exists():
    subprocess.run(['git', '-c', 'credential.helper=', 'clone', '--depth', '1', '--single-branch', '--branch', BRANCH, REPOSITORY, str(PROJECT)], check=True)
else:
    if not (PROJECT / 'tools/colab_runtime.py').is_file():
        raise RuntimeError('實驗程式目錄不完整，請重建執行階段。')
    current = subprocess.check_output(['git', '-C', str(PROJECT), 'branch', '--show-current'], text=True).strip()
    if current != BRANCH:
        raise RuntimeError(f'錯誤分支：{current}，應是 {BRANCH}。請使用全新執行階段。')
print('✅ 已鎖定移植測試分支：', BRANCH)
sys.path.insert(0, str(PROJECT / 'tools'))
import colab_runtime as runtime
runtime.install(globals())
```

```python
#@title ② 下載模型（實驗版）
INCLUDE_REWRITER = True #@param {type:"boolean"}
USE_DRIVE_CACHE = False #@param {type:"boolean"}
# 不下載 9B 時，請在前端取消「使用本地 9B 模型優化提示詞」。
runtime.download(globals(), include_rewriter=INCLUDE_REWRITER, use_drive_cache=USE_DRIVE_CACHE)
```

```python
#@title ③ 啟動 CLASSIC Python Slot 5070 Python 移植測試版
SAVE_TO_DRIVE = True #@param {type:"boolean"}
runtime.launch(globals(), save_to_drive=SAVE_TO_DRIVE, drive_folder='BananaMVClassicPythonSlot')
```
